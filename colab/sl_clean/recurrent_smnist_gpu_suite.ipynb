```python
import zipfile
from pathlib import Path

from google.colab import files

paths = [
    "results/sl_clean/gpu_c5_recurrent_smnist_panel.json",
    "results/sl_clean/gpu_streaming_smnist_row_panel.json",
    "results/sl_clean/gpu_b5_lstm_smnist_row_pareto_panel.json",
    "results/sl_clean/gpu_b6_esn_smnist_row_best_panel.json",
    "results/sl_clean/logs/gpu_c5_recurrent_smnist_panel.log",
    "results/sl_clean/logs/gpu_streaming_smnist_row_panel.log",
    "results/sl_clean/logs/gpu_b5_lstm_smnist_row_pareto_panel.log",
    "results/sl_clean/logs/gpu_b6_esn_smnist_row_best_panel.log",
]

missing = [p for p in paths if not Path(p).exists()]
assert not missing, f"Missing expected result files: {missing}"

zip_path = Path("results/sl_clean/recurrent_smnist_gpu_suite_results.zip")
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for path in paths:
        zf.write(path)
    for path in Path("results/sl_clean/colab_gpu_configs").glob("*.json"):
        zf.write(path)

print(f"Downloading {zip_path} ({zip_path.stat().st_size / 1024:.1f} KiB)")
files.download(str(zip_path))
```
