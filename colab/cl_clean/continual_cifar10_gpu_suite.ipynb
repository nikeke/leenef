# SSUP-II Phase C — Split-CIFAR-10 ConvNEF GPU suite

This notebook clones the Paper-II scaffold branch, installs the repository, runs the committed `cl_clean` GPU suite, and downloads the bundled artifacts.

```python
import shutil
import subprocess
import sys
from pathlib import Path

REPO_URL = "https://github.com/nikeke/leenef.git"
BRANCH = "paper-ii/cl-scaffold"  # Push this branch first, or replace with another pushed ref.
REPO_DIR = Path("/content/leenef")


def run_stream(cmd, *, cwd=None):
    proc = subprocess.Popen(
        cmd,
        cwd=cwd,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    )
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()
    if rc:
        raise subprocess.CalledProcessError(rc, cmd)


if REPO_DIR.exists():
    shutil.rmtree(REPO_DIR)

run_stream(["git", "clone", "--branch", BRANCH, "--single-branch", REPO_URL, str(REPO_DIR)])
run_stream([sys.executable, "-m", "pip", "install", "-q", "-e", ".[dev]"], cwd=REPO_DIR)
print(f"cloned {REPO_URL}@{BRANCH} to {REPO_DIR}")
```
