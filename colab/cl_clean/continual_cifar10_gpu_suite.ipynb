```python
import zipfile
from pathlib import Path

from google.colab import files

paths = [
    "results/cl_clean/gpu_c1_convnef_split_cifar10_panel.json",
    "results/cl_clean/gpu_c2_flat_nef_split_cifar10_panel.json",
    "results/cl_clean/logs/gpu_c1_convnef_split_cifar10_panel.log",
    "results/cl_clean/logs/gpu_c2_flat_nef_split_cifar10_panel.log",
]

missing = [p for p in paths if not Path(p).exists()]
assert not missing, f"Missing expected result files: {missing}"

zip_path = Path("results/cl_clean/continual_cifar10_gpu_suite_results.zip")
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for path in paths:
        zf.write(path)
    for path in Path("results/cl_clean/colab_gpu_configs").glob("*.json"):
        zf.write(path)

print(f"Downloading {zip_path} ({zip_path.stat().st_size / 1024:.1f} KiB)")
files.download(str(zip_path))
```
